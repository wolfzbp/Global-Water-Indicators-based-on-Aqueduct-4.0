# **Preparación y limpieza de datos**
___


En esta sección se describe el proceso de carga, exploración, filtrado y transformación del dataset utilizado en el análisis.

El objetivo es preparar un conjunto de datos consistente y adecuado para la construcción de visualizaciones que permitan analizar el riesgo hídrico en las provincias del Ecuador, y compararlo de manera global.

El dataset se obtuvo de: [Aqueduct 4.0 Current and Future Global Maps Data](https://www.wri.org/data/aqueduct-global-maps-40-data#download-form)

## **Carga de dataset**

```python
import pandas as pd
```

> Se cargo el dataset a un release de Github para poder leerlo desde una URL publica

```python
url = "https://github.com/wolfzbp/Global-Water-Indicators-based-on-Aqueduct-4.0/releases/download/v1.0/Aqueduct40_baseline_annual_y2023m07d05.csv"
df = pd.read_csv(url)
df
```

## **Seleccion de variables de interes**

```python
# clasificar columnas en categorias, y crear una lista de cada categoria con los nombres de las columnas correspondientes
identificadores = [
    "string_id", 
    "aq30_id",
    "name_0",
    "name_1",
    "area_km2"
]

riesgo_hidrologico = [
    "bws_score",   # Baseline water stress es el score de estrés hídrico de referencia
    "bwd_score",   # Water depletion es el score de agotamiento del agua
    "iav_score",   # Interannual variability es el score de variabilidad interanual
    "sev_score",   # Seasonal variability es el score de variabilidad estacional
    "drr_score"    # Drought risk es el score de riesgo de sequía
]

eventos_hidrologicos = [
    # "gtd_score",   # Groundwater table decline es el score de riesgo de declive de la tabla de agua subterránea, pero no se encuentra en el dataset
    "rfr_score",   # River flood risk es el score de riesgo de inundación fluvial
    "cfr_score"    # Coastal flood risk es el score de riesgo de inundación costera
]

calidad_agua = [
    "ucw_score",   # Untreated wastewater es el score de aguas residuales sin tratar
    "cep_score",   # Coastal eutrophication es el score de eutrofización costera
    "udw_score",   # No drinking water es el score de ausencia de agua potable
    "usa_score"    # No sanitation es el score de ausencia de saneamiento
]

por_industria = [
    "w_awr_agr_tot_score",   # Agriculture es el score de agricultura
    "w_awr_che_tot_score",   # Chemicals es el score de productos químicos
    "w_awr_con_tot_score",   # Construction materials es el score de materiales de construcción
    "w_awr_elp_tot_score",   # Electric power es el score de energía eléctrica
    "w_awr_fnb_tot_score",   # Food & beverage es el score de alimentos y bebidas
    "w_awr_min_tot_score",   # Mining es el score de minería
    "w_awr_ong_tot_score",   # Oil & gas es el score de petróleo y gas
    "w_awr_smc_tot_score",   # Semiconductor es el score de semiconductor
    "w_awr_tex_tot_score"    # Textile es el score de textil
]
```

```python
df_selected = df[identificadores + riesgo_hidrologico + eventos_hidrologicos + calidad_agua + por_industria]
df_selected.head()
```

## **Data Transformation: Restructuring**

> Pasar de wide format a long format para facilitar la visualizacion de las variables

```python
# Diccionario score -> categoria
mapa_categorias = {}

for col in riesgo_hidrologico:
    mapa_categorias[col] = "riesgo_hidrologico"

for col in eventos_hidrologicos:
    mapa_categorias[col] = "eventos_hidrologicos"

for col in calidad_agua:
    mapa_categorias[col] = "calidad_agua"

for col in por_industria:
    mapa_categorias[col] = "por_industria"

# Todas las columnas score
score_cols = list(mapa_categorias.keys())

# Supongamos que tu dataframe original se llama df_cleaned
# Aquí se conservan las demás columnas que NO son scores
id_cols = [col for col in df_cleaned.columns if col not in score_cols]

# Pasar a formato largo
df_long = df_cleaned.melt(
    id_vars=id_cols,
    value_vars=score_cols,
    var_name="score",
    value_name="value"
)

# Agregar la categoría
df_long["category"] = df_long["score"].map(mapa_categorias)

df_long
```

```python
mapa_nombres = {
    "bws_score": "Baseline water stress",
    "bwd_score": "Water depletion",
    "iav_score": "Interannual variability",
    "sev_score": "Seasonal variability",
    "drr_score": "Drought risk",
    "rfr_score": "River flood risk",
    "cfr_score": "Coastal flood risk",
    "ucw_score": "Untreated wastewater",
    "cep_score": "Coastal eutrophication",
    "udw_score": "No drinking water",
    "usa_score": "No sanitation",
    "w_awr_agr_tot_score": "Agriculture",
    "w_awr_che_tot_score": "Chemicals",
    "w_awr_con_tot_score": "Construction materials",
    "w_awr_elp_tot_score": "Electric power",
    "w_awr_fnb_tot_score": "Food & beverage",
    "w_awr_min_tot_score": "Mining",
    "w_awr_ong_tot_score": "Oil & gas",
    "w_awr_smc_tot_score": "Semiconductor",
    "w_awr_tex_tot_score": "Textile"
}

df_long["score_nombre"] = df_long["score"].map(mapa_nombres)
df_long
```
