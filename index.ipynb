## **Contexto del problema**

El agua es un recurso fundamental para el desarrollo humano, la producción industrial y la sostenibilidad ambiental. Sin embargo, su disponibilidad y uso no son uniformes en el territorio, incluso en países con alta riqueza hídrica como Ecuador.

En el contexto ecuatoriano, el uso del agua está directamente relacionado con actividades como el consumo doméstico, la producción agrícola y los procesos industriales. Estas demandas generan presión sobre los sistemas hídricos, especialmente cuando la disponibilidad del recurso es limitada o variable entre provincias.

Para analizar esta presión, se utiliza el concepto de **estrés hídrico**, que representa la relación entre la demanda de agua y el recurso disponible. En este proyecto, esta y otras dimensiones del riesgo no se analizan mediante valores absolutos, sino a través de **indicadores normalizados de manera técnica (scores entre 0 y 5)**, que permiten comparar territorios de manera consistente.

Sin embargo, el riesgo hídrico no depende únicamente de este indicador. También intervienen factores como la variabilidad del recurso en el tiempo, la ocurrencia de eventos extremos, la calidad del agua y su relación con diferentes sectores productivos. Estos elementos son representados mediante distintos indicadores que, en conjunto, permiten evaluar el riesgo de forma más integral en el territorio ecuatoriano.

Por esta razón, el análisis del riesgo hídrico en este proyecto se aborda desde un enfoque multidimensional, utilizando scores que sintetizan información compleja y facilitan la comparación entre provincias del Ecuador.
___

## **Motivación del análisis**

Aunque Ecuador es reconocido como un país con alta disponibilidad de recursos hídricos, esta percepción puede ocultar diferencias importantes a nivel territorial. No todas las provincias enfrentan las mismas condiciones en términos de presión sobre el recurso, exposición a eventos extremos o calidad del agua.

En muchos casos, el análisis del agua se limita a indicadores generales o promedios nacionales, lo que dificulta identificar zonas críticas o entender la naturaleza específica del riesgo en cada territorio.

Además, la información disponible suele ser compleja y técnica, lo que limita su uso en procesos de toma de decisiones si no se presenta de forma clara y accesible.

En este contexto, la visualización de datos aquí se convierte en una herramienta clave para:

- Comparar el riesgo hídrico entre provincias  
- Identificar patrones territoriales  
- Analizar distintas dimensiones del riesgo de manera integrada  
- Explorar la relación entre el recurso hídrico y sectores productivos  

Este proyecto busca transformar indicadores complejos en representaciones visuales comprensibles, permitiendo interpretar el riesgo hídrico en Ecuador de manera más clara y estructurada.

## **Preguntas que el proyecto busca responder**

El análisis se estructura a partir de un conjunto de preguntas que buscan entender el riesgo hídrico en Ecuador desde distintas dimensiones:


### 1. Contexto general

- ¿Cómo se posiciona Ecuador, a través del indicador de estrés hídrico, frente al contexto global?



### 2. Riesgo hidrológico

- ¿Qué provincias presentan mayores niveles de riesgo hidrológico?
- ¿Existen diferencias claras entre regiones del país (Costa, Sierra, Amazonía)?



### 3. Eventos hidrológicos

- ¿Qué provincias están más expuestas a inundaciones fluviales o costeras?
- ¿Se identifican patrones geográficos en la ocurrencia de estos eventos?



### 4. Calidad del agua

- ¿Qué provincias presentan mayores problemas relacionados con calidad del agua y acceso a servicios básicos?
- ¿Existe relación entre los indicadores de calidad del agua y los niveles de riesgo hidrológico?



### 5. Riesgo por industria

- ¿Qué sectores productivos presentan mayor riesgo hídrico en cada provincia?
- ¿Existen provincias donde múltiples sectores enfrentan altos niveles de riesgo?



### 6. Integración del análisis

- ¿Qué provincias concentran múltiples dimensiones de riesgo hídrico?
- ¿Se pueden identificar territorios críticos a partir de la combinación de indicadores?

## **Dataset utilizado**

El análisis se basa en el dataset **Aqueduct 4.0 Water Risk Atlas**, desarrollado por el *World Resources Institute (WRI)*, una de las principales fuentes globales para la evaluación del riesgo hídrico.

### **Fuente del dataset**

El dataset fue obtenido desde:

https://www.wri.org/data/aqueduct-water-risk-atlas

Este conjunto de datos integra información hidrológica, ambiental y socioeconómica, procesada mediante modelos globales, y es ampliamente utilizado en estudios académicos y toma de decisiones a nivel internacional.

### **Cobertura y estructura**

El dataset original tiene cobertura global a nivel subnacional. Para este proyecto, se realizó un filtrado para trabajar exclusivamente con **las provincias del Ecuador**, utilizando la variable `name_1` como identificador territorial.

Cada registro representa una unidad geográfica (provincia o combinación hidrológica-administrativa), asociada a distintos indicadores de riesgo hídrico.

### **Variables principales**

Se seleccionaron variables correspondientes a **identificación geográfica** y **scores de riesgo hídrico**, agrupadas en cuatro dimensiones:

#### 1. Riesgo hidrológico

- `bws_score` (estrés hídrico)  
- `bwd_score` (agotamiento del agua)  
- `iav_score` (variabilidad interanual)  
- `sev_score` (variabilidad estacional)  
- `drr_score` (riesgo de sequía)  

#### 2. Eventos hidrológicos

- `rfr_score` (inundaciones fluviales)  
- `cfr_score` (inundaciones costeras)  

#### 3. Calidad del agua

- `ucw_score` (aguas residuales sin tratar)  
- `cep_score` (eutrofización costera)  
- `udw_score` (acceso a agua potable)  
- `usa_score` (acceso a saneamiento)  

#### 4. Riesgo por industria

- `w_awr_agr_tot_score` (agricultura)  
- `w_awr_che_tot_score` (químicos)  
- `w_awr_con_tot_score` (construcción)  
- `w_awr_elp_tot_score` (energía eléctrica)  
- `w_awr_fnb_tot_score` (alimentos y bebidas)  
- `w_awr_min_tot_score` (minería)  
- `w_awr_ong_tot_score` (petróleo y gas)  
- `w_awr_smc_tot_score` (semiconductores)  
- `w_awr_tex_tot_score` (textil)  

### **Tipo de variables**

Las variables utilizadas en el análisis corresponden a:

- **Geográficas**: identificación de país y provincia (`name_0`, `name_1`)  
- **Numéricas**: scores de riesgo en escala de 0 a 5  
- **Categóricas**: variables construidas posteriormente (por ejemplo, región natural)  

### **Consideraciones del dataset**

Todos los indicadores utilizados están expresados como **scores normalizados en una escala de 0 a 5**, lo que permite comparar distintas dimensiones del riesgo hídrico entre territorios, aunque no representan valores absolutos de disponibilidad o consumo de agua.
